```python
import ROOT

# 1. Activamos el multi-threading para usar todos tus núcleos de CPU
ROOT.EnableImplicitMT()

# 2. Definimos el path y el árbol
path = "/lustre/LHCb/alejandro.rodriguez/script_emilio_hits/ntuple_signal_40114060.root"
tree_path = "VeloMultiTuple_73eaa531/Clusters"

# 3. Creamos el DataFrame (esto es instantáneo, no lee datos aún)
df = ROOT.RDataFrame(tree_path, path)

# 4. Definimos los histogramas que queremos (Lazy: solo se ejecutan al final)
# Formato: (Nombre, Título;EjeX;EjeY, Bins, Min, Max)
h_x = df.Histo1D(("h_x", "Distribución X;x [mm];Entries", 100, -50, 50), "x")
h_y = df.Histo1D(("h_y", "Distribución Y;y [mm];Entries", 100, -50, 50), "y")
h_z = df.Histo1D(("h_z", "Distribución Z;z [mm];Entries", 500, -500, 1000), "z")

# 5. Si quieres un histograma 2D (ej. hitmap del Velo)
h_xy = df.Histo2D(("h_xy", "Hitmap XY;x;y", 200, -50, 50, 200, -50, 50), "x", "y")

# 6. Ejecución real: al acceder a un histograma, ROOT dispara el bucle sobre los 10^9 eventos
# Esto ocurrirá una sola vez para TODOS los histogramas definidos arriba.
canvas = ROOT.TCanvas("c1", "Canvas", 1200, 400)
canvas.Divide(3, 1)

canvas.cd(1); h_x.Draw()
canvas.cd(2); h_y.Draw()
canvas.cd(3); h_z.Draw()

canvas.Draw() # Esto renderiza el canvas en la celda del notebook
```
